```python
from pathlib import Path
import shutil
from google.colab import files

def save_files(directory_name):
  directory = Path(directory_name)
  zip_name = f"{directory.name}_files"
  model_path = directory / "model"

  # Create zip archive
  archive_path = shutil.make_archive(zip_name, 'zip', model_path)

  # Download
  files.download(archive_path)
```

```python
# Experiment 1: BERT baseline with default settings for a fast 1-epoch reference run
output_dir = 'outputs/bert_baseline_default_1epoch'

!python models/bert_baseline.py --output_dir {output_dir} --epochs 1

save_files(output_dir)
```

```python
# Experiment 2: BERT with a smaller training subset to check whether faster iterations still learn anything
output_dir = 'outputs/bert_baseline_train10k'

!python models/bert_baseline.py --output_dir {output_dir} --epochs 1 --train_size 10000 --batch_size 8 --learning_rate 2e-5

save_files(output_dir)
```

```python
# Experiment 3: BERT with a larger training subset while keeping the run at 1 epoch
output_dir = 'outputs/bert_baseline_train45k'

!python models/bert_baseline.py --output_dir {output_dir} --epochs 1 --train_size 45000 --batch_size 8 --learning_rate 2e-5

save_files(output_dir)
```

```python
# Experiment 4: BERT with a lower learning rate to see if it improves minority-class recall
output_dir = 'outputs/bert_baseline_lr5e-6'

!python models/bert_baseline.py --output_dir {output_dir} --learning_rate 5e-6

save_files(output_dir)
```

```python
# Experiment 5: RoBERTa comparison run with the same 1-epoch budget
output_dir = 'outputs/roberta_baseline_1epoch'

!python models/bert_baseline.py --output_dir {output_dir} --model_name roberta-base

save_files(output_dir)
```
